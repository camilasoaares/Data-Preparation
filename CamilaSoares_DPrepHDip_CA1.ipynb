# Dealing with invalid datas
First we are going to convert these invalid datas into missing values, this way would be easier to clean our data set and subsequently handle to them. 
Except the value "five". Since code 5 represents Higher Education in your Data Dictionary, we can correct "five" to "5", rather than treating its as missing. 

```python
import numpy as np
data.loc[data['Age at enrollment'] == "UnKnown", 'Age at enrollment'] = np.nan
```

```python
print(data["Age at enrollment"].value_counts())
print(data["Age at enrollment"].unique())
```

```python
data.loc[data["Marital status"] == '?', "Marital status"] = np.nan
data.loc[data["Nacionality"] == '?', 'Nacionality'] = np.nan
data.loc[data["Displaced"] == '?', "Displaced"] = np.nan
data.loc[data["Tuition fees up to date"] == '?', "Tuition fees up to date"] = np.nan
```

```python
print("Marital status:")
print(data['Marital status'].unique())
print('\nNacionality:')
print(data['Nacionality'].unique())
print('\nDisplaced:')
print(data["Displaced"].unique())
print('\nTuition fees up to date')
print(data["Tuition fees up to date"].unique())
```

```python
data.loc[data["Curricular units 1st sem (credited)"] == "Na", "Curricular units 1st sem (credited)"] = np.nan
print(data["Curricular units 1st sem (credited)"].unique())
```

```python
data.loc[data["Father's occupation"] == "Five", "Father's occupation"] = "5"
print(data["Father's occupation"].unique())
print(data["Father's occupation"].value_counts())
```

```python
#Outliers 1st Semester
Q1 = data["Curricular units 1st sem (approved)"].quantile(0.25)
Q3 = data["Curricular units 1st sem (approved)"].quantile(0.75)
IQR = Q3 - Q1
print("Q1", Q1)
print("Q3", Q3)
print("IQR", IQR)
print("Lower limit:", Q1 - 1.5 * IQR)
print("Upper limit", Q3 + 1.5 * IQR)
```

```python
#Outliers 2nd Semester
Q1 = data["Curricular units 2nd sem (approved)"].quantile(0.25)
Q3 = data["Curricular units 2nd sem (approved)"].quantile(0.75)
IQR = Q3 - Q1
print("Q1", Q1)
print("Q3", Q3)
print("IQR", IQR)
print("Lower limit:", Q1 - 1.5 * IQR)
print("Upper limit", Q3 + 1.5 * IQR)
```

The IQR method was applied to investigate potential outliers in the numer of approved curricular units in both semesters. In the 1st semester, the IQR was 3, with an upper limit of 10.5. In the second semester, the IQR was 4, with an upper limit of 12. The results indicate that both variables contain unusually high values compared with their central distributions.
